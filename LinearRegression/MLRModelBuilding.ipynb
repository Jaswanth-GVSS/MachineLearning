```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

df=fetch_california_housing()

from sklearn.model_selection import train_test_split

X=df.data[:,0:8]
y=df.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
class MLR:
    def __init__(self):
        self.coeffs=None
        self.intercept=None

    def fit(self,X_train,y_train):

        X_train=np.asarray(X_train)
        y_train=np.asarray(y_train)

        X_train=np.insert(X_train,0,1,axis=1)
        

        betas = np.linalg.inv(np.dot(X_train.T, X_train)).dot(np.dot(X_train.T, y_train))
        self.intercept = betas[0]
        self.coeff = betas[1:]

    def predict(self,X_test):

        X_test = np.asarray(X_test)
        return np.dot(X_test, self.coeff) + self.intercept


        
```

```python
model=MLR()
model.fit(X_train,y_train)
```

```python
y_predict=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

r2_score(y_test,y_predict)
```

```python
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression  # <-- Actual Scikit-Learn model
from sklearn.metrics import r2_score

# Load and split data exactly like your code
df = fetch_california_housing()
X = df.data[:, 0:8]
y = df.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)

# 1. Train Scikit-Learn's actual Linear Regression
sklearn_model = LinearRegression()
sklearn_model.fit(X_train, y_train)

# 2. Predict
y_predict_sklearn = sklearn_model.predict(X_test)

# 3. Calculate R2 Score
sklearn_r2 = r2_score(y_test, y_predict_sklearn)

print(f"Scikit-Learn R2 Score : {sklearn_r2:.4f}")
print(f"Scikit-Learn Intercept: {sklearn_model.intercept_:.4f}")
```
