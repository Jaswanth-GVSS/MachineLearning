```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv(r'placement.csv')
```

```python
df
```

```python
X=df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
X.ndim
```

```python
y.ndim
```

```python
plt.scatter(X,y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
plt.scatter(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
y_pred
```

```python
X_test
```

```python
y_test
```

```python
lr.predict(np.array([5.88]).reshape(1,1))
```

```python
plt.scatter(X_train, y_train, color="blue", label="Actual data")
plt.plot(X_train, lr.predict(X_train), color="red", linewidth=2, label="Regression line")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

# How far off, on average, are the guesses (y_pred) from the truth (y_test)?
print(mean_absolute_error(y_test, y_pred))

# What is the overall accuracy score of the model?
print(r2_score(y_test, y_pred))
```
