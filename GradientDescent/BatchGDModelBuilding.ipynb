```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes

X,y=load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)

model=LinearRegression()
model.fit(X_train,y_train)

print(model.coef_)
print(model.intercept_)
```

```python
class BatchGD:
    def __init__(self,epoch,lr):
        self.lr=lr
        self.epoch=epoch
        self.coeff=None
        self.intercept=None
    
    def fit(self,X_train,y_train):

        self.intercept=0
        self.coeff=np.ones(X_train.shape[1])

        for i in range(self.epoch):

            y_predict=self.intercept+np.dot(X_train,self.coeff)
            intercept_der=-2*np.mean(y_train - y_predict)
            self.intercept=self.intercept-self.lr*intercept_der
           
            coeff_der=-2*np.dot(y_train - y_predict,X_train)/X_train.shape[0]
            self.coeff=self.coeff-self.lr*coeff_der
        
        print(self.coeff,self.intercept)
    
    def predict(self,X_test):
        return self.intercept+np.dot(X_test,self.coeff)


```

```python
sgd=BatchGD(6000,0.5)
sgd.fit(X_train,y_train)
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
y_pred=sgd.predict(X_test)
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test,y_pred))
```
