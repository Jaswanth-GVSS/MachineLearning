```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
from sklearn.datasets import make_regression

X,y=make_regression(n_samples=1000,n_features=5,random_state=42,noise=0.1)
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)

scaler=StandardScaler()
X_train_Scaled=scaler.fit_transform(X_train)
X_test_Scaled=scaler.transform(X_test)

sgdR=SGDRegressor(max_iter=1000,eta0=0.01,learning_rate='constant',random_state=42,tol=1e-3)

sgdR.fit(X_train_Scaled,y_train)
y_pred=sgdR.predict(X_test_Scaled)

print("Intercept : ",sgdR.intercept_)
print("coeffs : ",sgdR.coef_)
```
